# Carpet data exploration
Run after downloading the original MVTec AD carpet category. Use the project Python 3.11 kernel.
The notebook calls the package audit/split function and displays fitting images only. It does not tune models or thresholds.
Images: MVTec AD, MVTec Software GmbH, [CC BY-NC-SA 4.0](https://creativecommons.org/licenses/by-nc-sa/4.0/). Display resizing only. Clear outputs before committing.

```python
from pathlib import Path
import pandas as pd
from carpet_ad.data import prepare_data

root = Path.cwd()
if root.name == 'notebooks':
    root = root.parent
data_root = root / 'data/raw/carpet'
manifest = root / 'data/manifests/carpet_split.csv'
summary = prepare_data(data_root, manifest, root / 'reports/data_audit.md', seed=42)
print(summary)
rows = pd.read_csv(manifest)
rows.groupby(['original_split', 'split', 'defect_type']).size().rename('images')
```

```python
import matplotlib.pyplot as plt
from PIL import Image, ImageOps

samples = rows.loc[rows['split'] == 'fit', 'path'].head(6).tolist()
fig, axes = plt.subplots(2, 3, figsize=(9, 6))
for axis in axes.flat:
    axis.axis('off')
for axis, relative in zip(axes.flat, samples):
    with Image.open(data_root / relative) as image:
        axis.imshow(ImageOps.exif_transpose(image).convert('RGB'))
    axis.set_title(relative, fontsize=8)
fig.suptitle('Fitting samples — MVTec AD / MVTec Software GmbH / CC BY-NC-SA 4.0')
plt.tight_layout()
plt.show()
```
